## My rule and its reason codes

### Rule

My baseline rule identifies content that should be refreshed first.

Pages receive a higher score when they are older, have not been updated recently, have high search demand, have declining traffic, and already receive a meaningful number of impressions. These pages have the highest potential to benefit from a content refresh.

### Reason Code

- REFRESH_PRIORITY – Older content with declining performance and good search opportunity.

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
from pathlib import Path

# Load data
df = pd.read_csv("/content/content_refresh_anonymized.csv")

# Fill missing values
df["search_volume"] = df["search_volume"].fillna(0)
df["days_since_last_update"] = df["days_since_last_update"].fillna(0)
df["content_age_days"] = df["content_age_days"].fillna(0)
df["trend_pct"] = df["trend_pct"].fillna(0)
df["impressions_90d"] = df["impressions_90d"].fillna(0)

# Baseline Action Score
df["baseline_action_score"] = (
    0.35 * df["days_since_last_update"] +
    0.25 * df["content_age_days"] +
    0.20 * df["search_volume"] +
    0.10 * (-df["trend_pct"]) +
    0.10 * (df["impressions_90d"] / 100)
)

df["reason_code"] = "REFRESH_PRIORITY"
df["action"] = "Refresh Content"

ranked = df.sort_values("baseline_action_score", ascending=False)

Path("work/outputs").mkdir(parents=True, exist_ok=True)

ranked.to_csv(
    "work/outputs/baseline_action_score.csv",
    index=False
)

print(ranked[[
    "content_id",
    "baseline_action_score",
    "action",
    "reason_code"
]].head(20))
```
